```python
import os
from src.env.commons_env import HarvestCommonsEnv, MAP
```

```python
out_dir = 'output'
os.makedirs(out_dir, exist_ok=True)
```

```python
env = HarvestCommonsEnv(ascii_map=MAP['small'], num_agents=1, render=True)
env.reset()
env.render(mod="human")
# print initial observation
```

```python
# actions = {agent_id: env.action_space.sample() for agent_id in env.agents}
actions = {'agent-0': 0}
observations, rewards, dones, infos = env.step(actions)
env.render(mod='human')
```

```python
import matplotlib.pyplot as plt

# Plot observations for each agent
for agent_id, obs_dict in observations.items():
    obs = obs_dict["curr_obs"]
    plt.figure(figsize=(6, 6))
    plt.imshow(obs, interpolation="nearest")
    plt.title(f"Observation for {agent_id}")
    plt.axis('off')
    plt.show()
```

```python
print("\nRewards:")
print(rewards)
print("\nDones:")
print(dones)
```
